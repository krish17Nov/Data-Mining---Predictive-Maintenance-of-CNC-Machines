## Auto-ARIMA model for the time-series data

```python
from sklearn.preprocessing import LabelEncoder
df = pd.read_csv(r'D:\Project - Data Mining\Project\Data-Mining---Predictive-Maintenance-of-CNC-Machines\Local Repository\Data-Mining---Predictive-Maintenance-of-CNC-Machines\Pre-processing\Tool-Wear-Detection\experiment_01.csv')
le = LabelEncoder()
df['Machining_Process_Encoded'] = le.fit_transform(df['Machining_Process'])
```

```python
sequence = df['Machining_Process_Encoded'].values
```

```python
from statsmodels.tsa.arima.model import ARIMA

model = ARIMA(sequence, order = (2,0,1))
model_fit = model.fit()

next_step_encoded = model_fit.forecast(steps=1)[0]

next_step_label = le.inverse_transform([int(round(next_step_encoded))])[0]
print("Next predicted machining step:", next_step_label)
```
